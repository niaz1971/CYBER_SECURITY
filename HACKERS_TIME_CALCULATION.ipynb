```python
from decimal import Decimal, getcontext

# High precision so huge keyspaces (like 2^128) don't lose accuracy
getcontext().prec = 60

# Given values
key_bits = 2  # Encryption key length in bits
keys_per_second = Decimal(10**12)  # Attacker speed: 10^12 keys/sec
seconds_per_year = Decimal(60 * 60 * 24 * 365)  # 31,536,000 = 3.1536e7

# Step 1: Total possible keys = 2^key_bits
total_keys = Decimal(2**key_bits)

# Step 2: Worst-case time in seconds (last key is the correct match)
worst_case_seconds = total_keys / keys_per_second

# Step 3: Average-case time in seconds (~50% of keys tested)
average_case_seconds = worst_case_seconds / Decimal(2)

# Step 4: Convert worst-case and average-case times into years
worst_case_years = worst_case_seconds / seconds_per_year
average_case_years = average_case_seconds / seconds_per_year

# Display step-by-step solution
print("=== BRUTE-FORCE SOLUTION ===")
print(f"Encryption bit key       = {key_bits} bit")
print(f"Total Keys (2^{key_bits})     = {total_keys:,.0f} (~{total_keys:.2e})")
print(f"Attacker keys per second = {keys_per_second:,.0f} ({keys_per_second:.0e})")
print("-" * 60)
print(f"Worst-Case Time (sec)    = {worst_case_seconds:.2e} seconds")
print(f"Average-Case Time (sec)  = {average_case_seconds:.2e} seconds")
print("-" * 60)
print(f"Seconds per year         = {seconds_per_year:,.0f} ({seconds_per_year:.4e})")
print(
    f"Worst-Case Time (years)  = {worst_case_years:.2e} years "
    f"(~{worst_case_years / Decimal('1e18'):.2f} quintillion years)"
)
print(
    f"Average-Case Time (years)= {average_case_years:.2e} years "
    f"(~{average_case_years / Decimal('1e18'):.2f} quintillion years)"
)
```
