```python
def vigenere_cipher(text: str, key: str, mode: str = "encrypt") -> str:
    """Encrypt or decrypt text using a Vigenere cipher."""
    if mode not in ("encrypt", "decrypt"):
        raise ValueError("mode must be 'encrypt' or 'decrypt'")

    clean_key = [ch.upper() for ch in key if ch.isalpha()]
    if not clean_key:
        raise ValueError("key must contain at least one alphabetic letter")

    result = []
    key_idx = 0

    for char in text:
        if char.isalpha():
            base = ord("A") if char.isupper() else ord("a")
            shift = ord(clean_key[key_idx % len(clean_key)]) - ord("A")
            if mode == "decrypt":
                shift = -shift

            shifted = (ord(char) - base + shift) % 26 + base
            result.append(chr(shifted))
            key_idx += 1
        else:
            result.append(char)  # Preserves spaces and punctuation

    return "".join(result)


if __name__ == "__main__":
    message = "HAMZA LEARNING"
    key = "GOOD"

    encrypted = vigenere_cipher(message, key, mode="encrypt")
    decrypted = vigenere_cipher(encrypted, key, mode="decrypt")

    print(f"Original:  {message}")
    print(f"Key:       {key}")
    print(f"Encrypted: {encrypted}")
    print(f"Decrypted: {decrypted}")
```

```python
def shift_cipher(text: str, shift: int, mode: str = "encrypt") -> str:
    """Encrypt or decrypt text using a shift (Caesar) cipher."""
    if mode == "decrypt":
        shift = -shift
    elif mode != "encrypt":
        raise ValueError("mode must be 'encrypt' or 'decrypt'")

    result = []
    for char in text:
        if char.isalpha():
            base = ord("A") if char.isupper() else ord("a")
            shifted = (ord(char) - base + shift) % 26 + base
            result.append(chr(shifted))
        else:
            result.append(char)  # Keeps spaces and punctuation unchanged

    return "".join(result)


if __name__ == "__main__":
    message = "subscribe to channel study table"
    key = 5

    encrypted = shift_cipher(message, key, mode="encrypt")
    decrypted = shift_cipher(encrypted, key, mode="decrypt")

    print(f"Original:  {message}")
    print(f"Encrypted: {encrypted}")
    print(f"Decrypted: {decrypted}")
```
